```python
class Smartphone:
    # 1. Metodo __init__ para inicializar atributos (constructor)
    def __init__(self, marca, modelo, capacidadAlmacenamiento, sistemaOperativo):
        self.marca = marca
        self.modelo = modelo
        self.capacidadAlmacenamiento = capacidadAlmacenamiento
        self.sistemaOperativo = sistemaOperativo

    # 2. Definición de funciones (métodos)
    def llamar(self, contacto):
        print(f"Llamando a {contacto} desde mi {self.modelo}...")

    def tomarFoto(self):
        print(f"¡Click! Foto capturada con el {self.marca} {self.modelo}.")

    def instalarApp(self, nombreApp):
        print(f"Instalando {nombreApp} en el sistema {self.sistemaOperativo}...")

    def mostrar_info(self):
        print(f"Smartphone: {self.marca} {self.modelo}")
        print(f"Almacenamiento: {self.capacidadAlmacenamiento}GB | SO: {self.sistemaOperativo}")

# --- Creación de objetos (instancias) ---
celular1 = Smartphone("Google", "Pixel 8", 128, "Android 14")
celular2 = Smartphone("Apple", "iPhone 15 Pro", 256, "iOS 17")

# --- Llamadas a las funciones ---
print("--- Celular 1 ---")
celular1.mostrar_info()
celular1.llamar("Karol")
celular1.instalarApp("Instagram")

print("\n--- Celular 2 ---")
celular2.mostrar_info()
celular2.tomarFoto()
celular2.instalarApp("WhatsApp")

print("\n-------------------------")
print("Programa terminado")
print("Creador de software por Karol Castillo NL = 9")
```
